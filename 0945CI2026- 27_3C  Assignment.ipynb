5.Create a Pandas DataFrame with 2 columns: 'Category' and 'Value'. Fill the 'Category' column with random categories ('A', 'B', 'C') and the 'Value' column with random integers. Group the DataFrame by 'Category' and compute the sum and mean of 'Value' for each category

```python
import pandas as pd

df = pd.DataFrame({
    'Category': ['A', 'B', 'A', 'C', 'B', 'A', 'C', 'B'],
    'Value': [10, 20, 15, 30, 25, 5, 40, 10]
})

result = df.groupby('Category')['Value'].agg(['sum', 'mean'])

print("DataFrame:")
display(df)

print("Sum and mean by category:")
display(result)
```

6. Create a Pandas DataFrame with 3 columns: 'Product', 'Category', and 'Sales'. Fill the DataFrame with random data. Group the DataFrame by 'Category' and compute the total sales for each category.



```python
import pandas as pd

df = pd.DataFrame({
    'Product': ['Pen', 'Notebook', 'Pencil', 'Bag', 'Ruler', 'Book'],
    'Category': ['Stationery', 'Stationery', 'Stationery',
                 'Accessories', 'Stationery', 'Stationery'],
    'Sales': [100, 250, 80, 700, 60, 300]
})

total_sales = df.groupby('Category')['Sales'].sum()

print("DataFrame:")
display(df)

print("Total sales by category:")
display(total_sales)
```

7. Merging DataFrames

Create two Pandas DataFrames with a common column. Merge the DataFrames using the common column.



```python
import pandas as pd

df1 = pd.DataFrame({
    'StudentID': [1, 2, 3],
    'Name': ['Anu', 'Ravi', 'Maya']
})

df2 = pd.DataFrame({
    'StudentID': [1, 2, 3],
    'Marks': [85, 90, 78]
})

merged_df = pd.merge(df1, df2, on='StudentID')

display(merged_df)
```

8. MultiIndex DataFrame

Create a Pandas DataFrame with a MultiIndex (hierarchical index). Perform some basic indexing and slicing operations on the MultiIndex DataFrame.

```python
import pandas as pd

# Create a MultiIndex with two levels: Department and Employee
index = pd.MultiIndex.from_tuples(
    [('Sales', 'Anu'),
     ('Sales', 'Ravi'),
     ('IT', 'Maya'),
     ('IT', 'Arun')],
    names=['Department', 'Employee']
)

df = pd.DataFrame(
    {'Salary': [30000, 35000, 40000, 45000]},
    index=index
)

print("MultiIndex DataFrame:")
display(df)

# Select all employees in Sales
display(df.loc['Sales'])

# Select one employee using both index levels
display(df.loc[('IT', 'Maya')])

# Select rows from Sales through IT
display(df.loc['Sales':'IT'])
```

9. Create a Pandas DataFrame with MultiIndex consisting of 'Category' and 'SubCategory'. Fill the DataFrame with random data and compute the sum of values for each 'Category' and 'SubCategory'.

```python
import pandas as pd

df = pd.DataFrame({
    'Category': ['A', 'A', 'A', 'B', 'B', 'B'],
    'SubCategory': ['X', 'X', 'Y', 'X', 'Y', 'Y'],
    'Value': [10, 15, 20, 5, 12, 8]
})

# Make Category and SubCategory the hierarchical index
multi_df = df.set_index(['Category', 'SubCategory'])

print("MultiIndex DataFrame:")
display(multi_df)

print("Sum for each Category and SubCategory:")
display(multi_df.groupby(level=['Category', 'SubCategory'])['Value'].sum())
```

10. Applying Functions

Create a Pandas DataFrame with 3 columns and 5 rows filled with random integers. Apply a function that doubles the values of the DataFrame.



```python
import pandas as pd

df = pd.DataFrame({
    'A': [2, 4, 6, 8, 10],
    'B': [1, 3, 5, 7, 9],
    'C': [10, 20, 30, 40, 50]
})

doubled_df = df.apply(lambda column: column * 2)

print("Original DataFrame:")
display(df)

print("DataFrame with values doubled:")
display(doubled_df)
```

11. Create a Pandas DataFrame with 3 columns and 6 rows filled with random integers. Apply a lambda function to create a new column that is the sum of the existing columns.



```python
import pandas as pd

df = pd.DataFrame({
    'A': [2, 4, 6, 8, 10, 12],
    'B': [1, 3, 5, 7, 9, 11],
    'C': [10, 20, 30, 40, 50, 60]
})

df['RowSum'] = df.apply(lambda row: row['A'] + row['B'] + row['C'], axis=1)

display(df)
```
